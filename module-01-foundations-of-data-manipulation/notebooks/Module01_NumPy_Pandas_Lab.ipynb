## Part 1 — NumPy arrays

### Learning Objectives
- Create arrays and read their attributes
- Use arithmetic, aggregation and broadcasting
- Select data with slicing, boolean masks and fancy indexing

### Concept
A NumPy array stores numbers of **one data type** in a contiguous block of memory. That makes it compact and lets NumPy apply an operation to every element at once (*vectorisation*).

### Guided Practice

```python
import numpy as np

daily_steps = np.array([4200, 8100, 6500, 12000, 9300, 3000, 7600])
print("shape:", daily_steps.shape, "| dtype:", daily_steps.dtype)
print("Total steps this week:", daily_steps.sum())
print("Average per day:", daily_steps.mean().round(1))
print("Days above 8000:", daily_steps[daily_steps > 8000])
```

```python
# 2D arrays, axis and broadcasting
week = np.arange(1, 15).reshape(2, 7)   # 2 weeks x 7 days
print(week)
print("Per-day totals (axis=0):", week.sum(axis=0))
print("Per-week totals (axis=1):", week.sum(axis=1))
print("Add a bonus to each weekday:\n", week + np.array([0, 0, 0, 0, 0, 5, 5]))
```

### Concept
A **Series** is a labelled 1D array. A **DataFrame** is a table of Series that share one index. Selection uses `df['col']`, `df.loc[row_condition, columns]` (labels) and `df.iloc[rows, cols]` (positions).

## Part 3 — Data input/output

### Learning Objectives
- Read files with column and row options
- Write CSV, Excel and JSON files

### Concept
`pd.read_csv()` has many options: `usecols` (columns to keep), `nrows` (rows to read), `index_col` (row labels) and `header` (which row holds the names). Writers mirror readers: `to_csv()`, `to_excel()`, `to_json()`.

### Guided Practice

```python
sales = load_data("retail_sales.csv", usecols=["date", "region", "category", "revenue"], parse_dates=["date"])
monthly = sales.groupby(sales["date"].dt.month)["revenue"].sum().round(0)
monthly.to_csv("monthly_revenue.csv", header=["revenue"])
print(pd.read_csv("monthly_revenue.csv", index_col=0).head())
```
