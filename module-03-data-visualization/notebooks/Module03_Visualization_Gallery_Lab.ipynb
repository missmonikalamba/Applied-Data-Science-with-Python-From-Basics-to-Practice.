### Learning Objectives

`fig, ax = plt.subplots()` gives you a Figure (the canvas) and an Axes (one plot). Call drawing methods on `ax` and finish every chart with a title, axis labels and, when needed, a legend.

### Guided Practice

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sales = load_data("retail_sales.csv", parse_dates=["date"])
monthly = sales.groupby(sales["date"].dt.month)["revenue"].sum()

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(monthly.index, monthly.values, marker="o", color="#0E7C86")
ax.set_title("Monthly revenue, 2024")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue")
ax.grid(alpha=0.3)
plt.show()
```

## Part 2 — Subplots, annotations, categorical and time series charts

### Learning Objectives
- Build a 2×2 grid of charts
- Annotate the key point
- Plot a time series with formatted dates

### Concept
`plt.subplots(nrows, ncols)` returns an array of Axes. `ax.annotate(text, xy=point, xytext=label_position, arrowprops=...)` draws attention to one point. Plot datetimes directly and format the axis with `matplotlib.dates`.

### Guided Practice

```python
import matplotlib.dates as mdates
weekly = sales.set_index("date")["revenue"].resample("W").sum()
fig, ax = plt.subplots(figsize=(9, 3.5))
ax.plot(weekly.index, weekly.values, color="#0E7C86")
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
peak = weekly.idxmax()
ax.annotate("Holiday peak", xy=(peak, weekly.max()), xytext=(peak - pd.Timedelta(days=120), weekly.max()),
            arrowprops=dict(arrowstyle="->"))
ax.set_title("Weekly revenue, 2024")
plt.show()
```
